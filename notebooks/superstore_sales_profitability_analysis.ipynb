## 7. Function Overview: Summary
This function generates a KPI summary table by any grouping variable (such as state, region, or category). It calculates totals, contribution percentages, profit margin, and average discount, then formats the results neatly for presentation.

Inputs:

1. df = dataset with columns like sales, profit, quantity, order_id, discount.
2. group_var = column to group by (e.g., state).
3. metrics = list of metrics to calculate (sales, profit, quantity, orders, discount).
4. include_percentages = whether to add contribution percentages and profit margin.
5. sort_by = column to sort results by (default = first metric).
6. ascending = sort order (default = largest first).

Outputs:

A clean summary table with totals, percentages, margins, and average discount.
All values are formatted for readability (currency, commas, percentages).

Why it’s useful:
Saves time by avoiding repeated groupby code.
Flexible: you choose metrics, sorting, and whether to include percentages.
Produces professional, portfolio‑ready tables that recruiters and non‑technical readers can understand.

```python
# Summary function

def summary(df, group_var, metrics=['sales','profit'], include_percentages=True, sort_by=None, ascending=False):
    agg_dict = {}
    
    # Build aggregation dictionary in the order metrics are passed
    for metric in metrics:
        if metric == 'sales':
            agg_dict['total_sales'] = ('sales','sum')
        elif metric == 'profit':
            agg_dict['total_profit'] = ('profit','sum')
        elif metric == 'quantity':
            agg_dict['total_quantity'] = ('quantity','sum')
        elif metric == 'orders':
            agg_dict['order_count'] = ('order_id','nunique')
            agg_dict['line_item_count'] = ('order_id','count')
        elif metric == 'discount':
            agg_dict['avg_discount'] = ('discount','mean')
    
    result = df.groupby(group_var).agg(**agg_dict).reset_index()
    
    # Percentages and margin
    if include_percentages:
        if 'total_sales' in result.columns:
            result['sales_percentage'] = result['total_sales'] / result['total_sales'].sum() * 100
        if 'total_profit' in result.columns:
            result['profit_percentage'] = result['total_profit'] / result['total_profit'].sum() * 100
        if 'total_quantity' in result.columns:
            result['quantity_percentage'] = result['total_quantity'] / result['total_quantity'].sum() * 100
        if 'order_count' in result.columns:
            result['order_count_percentage'] = result['order_count'] / result['order_count'].sum() * 100
        if 'line_item_count' in result.columns:
            result['line_item_percentage'] = result['line_item_count'] / result['line_item_count'].sum() * 100
        if 'total_sales' in result.columns and 'total_profit' in result.columns:
            result['profit_margin'] = result['total_profit'] / result['total_sales'] * 100
    
    # Sorting
    if sort_by is None:
        sort_by = list(agg_dict.keys())[0]  # default: first metric chosen
    result = result.sort_values(sort_by, ascending=ascending)
    
    # Formatting
    for col in ['total_sales','total_profit']:
        if col in result.columns:
            result[col] = result[col].apply(lambda x: f'${x:,.2f}')
    for col in ['total_quantity','order_count','line_item_count']:
        if col in result.columns:
            result[col] = result[col].apply(lambda x: f'{x:,}')
    if 'avg_discount' in result.columns:
        result['avg_discount'] = result['avg_discount'].apply(lambda x: f'{x:.2f}%')
    if include_percentages:
        for col in ['sales_percentage','profit_percentage','quantity_percentage',
                    'order_count_percentage','line_item_percentage','profit_margin']:
            if col in result.columns:
                result[col] = result[col].apply(lambda x: f'{x:.2f}%')
    
    # Reorder columns so metrics appear in the same order as passed
    ordered_cols = [group_var] + list(agg_dict.keys())
    other_cols = [c for c in result.columns if c not in ordered_cols]
    result = result[ordered_cols + other_cols]
    
    return result
```

## 8.2 Profit Summary
The sustainability of any business relies heavily on its profit performance. Profit is not just a measure of financial success; it is the foundation for growth, resilience, and long‑term viability. Therefore, conducting a thorough profit analysis is critical to understanding operational efficiency, identifying opportunities for improvement, and ensuring the business remains competitive in the market.

```python
# Profit summary

gross_profits = df.loc[df['profit'] > 0, 'profit'].sum() 
total_losses = df.loc[df['profit'] < 0, 'profit'].sum()
net_profit = df['profit'].sum()

potential_profit = gross_profits + abs(total_losses)

discount_impact = abs(total_losses) / potential_profit * 100

profit_summary = pd.DataFrame({
    'gross_profits': [gross_profits],
    'Total Losses': [total_losses],
    'Net Profit': [net_profit],
    'Potential Profit (No Losses)': [potential_profit],
    'Discount Impact': [discount_impact],
    'Net Margin': [net_profit / df['sales'].sum() * 100]
})

for col in ['gross_profits', 'Total Losses', 'Net Profit', 'Potential Profit (No Losses)']:
    profit_summary[col] = profit_summary[col].apply(lambda x: f'${x:,.2f}')

for col in ['Discount Impact', 'Net Margin']:
    profit_summary[col] = profit_summary[col].apply(lambda x: f'{x:,.2f}%')

print(f'\n{profit_summary}')
```

### Interpretation
The business generated a gross profit of approximately USD 449,300.33 from profitable transactions. However, losses totaling USD 157,026.87 reduced this figure substantially, leaving a net profit of USD 292,273.46. This means that about 35% of total profit was eroded by losses, highlighting the impact of unprofitable transactions on overall performance.
The net profit margin of 12.56% indicates that for every USD 100 in sales revenue, only about USD 12.56 remains as profit after accounting for losses. While the business is profitable overall, the margin suggests moderate efficiency.

If there had been no losses, the business could have achieved a potential profit of USD 606,327.20.
The Discount Impact of 25.90% reveals that more than a quarter of potential earnings were wiped out by discounts, returns, or other loss inducing factors.

## 8.8 Monthly Sales Trend
Sales are the lifeblood of every business, as they determine whether the company is progressing or regressing in its performance.

```python
# Monthly sales trend

df['year'] = df['order_date'].dt.year
df['month'] = df['order_date'].dt.month

monthly_sales_trend = (
    df.groupby(['year', 'month'])['sales'].sum()
        .reset_index()
)

monthly_sales_trend['Yearly_Total'] = (
    monthly_sales_trend.groupby('year')['sales']
        .transform('sum')
)

monthly_sales_trend['Percentage'] = (monthly_sales_trend['sales'] / monthly_sales_trend['Yearly_Total']) * 100

print(monthly_sales_trend)
```

### Interpretation
The monthly sales trend analysis reveals clear seasonality. Across all years, November and December consistently account for the highest share of annual sales, highlighting the impact of holiday demand. In contrast, January and February contribute the least, reflecting post‑holiday slowdowns. Mid‑year months such as March, May, and September show intermittent peaks.

## 8.9 Monthly Profit Trend

```python
# Monthly profit trend

monthly_profit_trend = (
    df.groupby(['year', 'month'])['profit'].sum()
        .reset_index()
)

monthly_profit_trend['Yearly_Total'] = (
    monthly_profit_trend.groupby('year')['profit']
        .transform('sum')
)

monthly_profit_trend['Percentage'] = (monthly_profit_trend['profit'] / monthly_profit_trend['Yearly_Total']) * 100

print(monthly_profit_trend)
```

## 9.3 Sub-Category by Quantity, Sales and Profit
It is essential to compare demand volume with sales and profit to identify sub‑categories that sell in high quantities but generate low revenue, profit and those with low quantities yet high sales, often premium products. This distinction highlights volume‑driven versus value‑driven segments.

```python
# Sub-categories by quantity, sales and profit with Avg sales and profit per unit

sub_category = summary(
    df,
    group_var='sub_category',
    metrics=['quantity', 'sales', 'profit'],
    include_percentages=False,
    sort_by='total_quantity'
)

sub_category['sales_value'] = sub_category['total_sales'].str.replace('[\$,]', '', regex=True).astype(float)
sub_category['quantity_value'] = sub_category['total_quantity'].str.replace(',', '', regex=True).astype(float)

sub_category['profit_value'] = sub_category['total_profit'].str.replace('[\$,]', '', regex=True).astype(float)
sub_category['quantity_value'] = sub_category['total_quantity'].str.replace(',', '', regex=True).astype(float)

sub_category['avg_sales_per_unit'] = (sub_category['sales_value'] / sub_category['quantity_value']).apply(lambda x: f'${x:,.2f}')
sub_category['avg_profit_per_unit'] = (sub_category['profit_value'] / sub_category['quantity_value']).apply(lambda x: f'${x:,.2f}')

sub_category = sub_category.drop(columns=['sales_value', 'profit_value','quantity_value'])

print(f'\nSub-Category by Quantity, Sales and Profit (with Avg Sales and Profit per Unit):\n\n{sub_category}')
```

```python
import calendar

monthly_customers = (
    df.groupby([df['order_date'].dt.year.rename('year'),
                df['order_date'].dt.month.rename('month')])['customer_name']
      .nunique()
      .reset_index(name='customer_count')
)

monthly_customers['percentage'] = (
    monthly_customers.groupby('year')['customer_count']
    .transform(lambda x: x / x.sum() * 100)
)

monthly_customers['month_name'] = monthly_customers['month'].transform(lambda x: calendar.month_abbr[x])

pivot_table = monthly_customers.pivot(index='year', columns='month_name', values='percentage')

print(pivot_table)
```

## 13.3 Creation of Customer Category Column
The introduction of a Customer Category column is a critical step in enhancing customer analysis. This column is designed to classify customers based on two key dimensions: their longevity with the business and the total number of unique orders they have placed. By combining these factors, the column provides a structured way to distinguish between one‑time buyers, occasional purchasers, repeat buyers, loyal customers, and dormant accounts.

```python
# Creation of Customer Category Column

customer_purchase_span = df.groupby('customer_name')['order_date'].agg(['min','max','count'])
customer_purchase_span['loyalty_days'] = (customer_purchase_span['max'] - customer_purchase_span['min']).dt.days
customer_purchase_span.rename(columns={'count':'purchase_count'}, inplace=True)

def segment_customer(row):
    if row['purchase_count'] == 1:
        return 'One-time Buyer'
    elif row['purchase_count'] <= 3:
        return 'Occasional Buyer'
    elif row['purchase_count'] <= 10:
        return 'Repeat Buyer'
    elif row['purchase_count'] > 10 and row['loyalty_days'] > 1000:
        return 'Loyal Customer'
    else:
        return 'Dormant Customer'

customer_purchase_span['customer_category'] = customer_purchase_span.apply(segment_customer, axis=1)

customer_purchase_span = customer_purchase_span.reset_index()

df = df.merge(customer_purchase_span[['customer_name','customer_category']], 
              on='customer_name', 
              how='left')

print(df[['customer_name','segment','customer_category']].head(10))
```

## 14.6 Creation of Loss Order Function
This function is designed to group categorical column(s) by total sales, total profit, average discount, and count of orders. By structuring the data this way, it provides a clear view of how discounting practices affect profitability across different dimensions. The insights gained make it easier to identify where high discounts are eroding margins and to compare performance across categories, regions, or states.

```python
# Loss orders function

def loss_orders_summary(loss_orders, groupby_cols):
    summary = (loss_orders
               .groupby(groupby_cols)
               .agg({
                   'sales': 'sum',
                   'profit': 'sum',
                   'discount': 'mean',
                   'order_id': 'count'
               })
               .rename(columns={'order_id': 'loss_order_count'})
               .reset_index()
               .sort_values('discount', ascending=False))
    
    return summary
```

```python
# OLS Regression

import statsmodels.api as sm

product_dummies = pd.get_dummies(df['sub_category'], drop_first=True)
region_dummies = pd.get_dummies(df['region'], drop_first=True)

X = pd.concat([df[['sales', 'quantity', 'discount']], product_dummies, region_dummies], axis=1)

X = sm.add_constant(X)

X = X.astype(float)

y = df['profit'].astype(float)

model = sm.OLS(y, X).fit()

print(model.summary())
```

### Interpretation
The OLS regression model was statistically significant overall (F = 234.7, p < 0.001) and explained approximately 33.7% of the variation in profit (R² = 0.337; adjusted R² = 0.335). This indicates that sales, quantity, discount, product sub-category, and region collectively provide meaningful explanatory power for profitability, although a substantial proportion of profit variation remains unexplained by the model.

Sales had a positive and statistically significant association with profit (β = 0.186, p < 0.001), while quantity showed a statistically significant negative association (β = −3.277, p < 0.001). Most notably, discount had a large negative and statistically significant coefficient (β = −274.652, p < 0.001), reinforcing the descriptive finding that higher discounting is strongly associated with lower profitability.

Product sub-category effects varied considerably. Binders (β = 61.35, p < 0.001) and Copiers (β = 406.80, p < 0.001) showed significant positive associations with profit relative to the reference sub-category, while Bookcases, Chairs, Machines, Storage, Supplies, and Tables showed significant negative associations. Machines (β = −223.91) and Tables (β = −137.42) were among the strongest negative sub-category effects. Several other sub-categories were not statistically significant, indicating insufficient evidence of a distinct profitability effect after controlling for the other variables.

Regional coefficients were also statistically significant for East, South, and West relative to the reference region, although these regression results should be interpreted differently from the descriptive regional rankings because the regression controls for other factors such as sales, quantity, discount, and product mix.

Overall, the regression supports the broader finding that profitability is influenced by a combination of sales volume, discounting, product mix, quantity, and regional factors rather than sales alone. The strong negative association between discount and profit and the substantial differences across sub-categories suggest that improving profitability requires greater control over discounting and closer management of low-margin product lines.

The model should nevertheless be interpreted with caution. Its R² of 0.337 indicates that considerable variation in profit remains unexplained, while the residual diagnostics indicate substantial departures from normality. The large condition number also suggests potential scaling or multicollinearity issues. Therefore, the regression is best treated as evidence of statistical associations and explanatory patterns rather than causal effects

## 16. Key Findings

1. Overall Profitability
The business generated USD 449,300.33 from profitable transactions, while loss-making transactions contributed USD 157,026.87 in losses, resulting in a net profit of USD 292,273.46. Approximately 1 in 5 orders resulted in a net loss.

2. Discounting & Profitability
Loss-making orders received approximately six times more discount, on average, than profitable orders, while discount and profit showed a strong negative correlation. This indicates that aggressive discounting is a major potential contributor to profitability challenges.

3. Product & Category Performance
Technology generated the highest share of revenue and profit despite fewer orders and lower quantities, while Tables, Bookcases, and Supplies negatively affected profitability. Tables were particularly notable for generating above-average revenue while recording negative profit.

4. High-Margin Products
Chairs and Phones were major revenue drivers, while Copiers, Labels, and Envelopes achieved exceptionally high profit margins. Binders recorded the highest quantity sold and were also among the products receiving the highest discounts.

5. Customer Value & Retention
Loyal customers represented the largest customer group and generated the majority of revenue and profit. Technology also had the shortest average customer revisit period, indicating stronger repeat-purchase behavior.

6. Geographic Concentration & Risk
The United States accounted for approximately 98.7% of revenue and 97.9% of profit, with the West region leading revenue, profit, margin, and order volume. California was the leading state by revenue and profit, while Texas was the leading loss-making state.

7. Geographic Discount Pattern
Six states recorded losses despite generating substantial revenue and also had above-average discounts. The Central region and Office Supplies recorded the highest average discounts, suggesting that discounting deserves closer geographic and category-level monitoring.

8. Seasonality

## 17. Key Business Conclusions

1. Sales do not always translate into profitability

Technology and Office Supplies are major contributors to sales and profitability, but strong revenue does not always result in strong profit. Furniture, in particular, contains sub-categories where substantial sales are accompanied by low or negative profitability, highlighting the importance of evaluating revenue together with margin and discount levels.

2. Profitability varies considerably across regions

The West and East regions are strong contributors to overall sales and profit, while the South performs comparatively weaker in overall business contribution. However, the regression results indicate that regional differences remain statistically significant even after controlling for sales, quantity, discount, and product mix, showing that descriptive rankings and regression-based effects provide different perspectives on regional performance.

3. Customer segments differ in scale and profitability efficiency

Consumer is the largest segment by sales, profit, quantity, and order volume, making it the primary driver of business activity. Corporate generates substantial revenue and maintains a higher profit margin than Consumer. Home Office has the lowest sales and absolute profit but records the highest profit margin, demonstrating that smaller revenue contribution does not necessarily mean weaker profitability efficiency.

4. Discounting is a major profitability concern

Higher discount levels are strongly associated with lower profitability. Loss-making orders receive substantially higher discounts than profitable orders, while the regression also identifies discount as one of the strongest negative and statistically significant predictors of profit. This suggests that aggressive discounting may erode margins without generating sufficient additional value to compensate for the reduction in profit.

5. Sales volume alone is not a reliable measure of performance

Higher quantities do not necessarily produce higher profits. The regression identifies quantity as having a statistically significant negative association with profit after controlling for sales, discount, product sub-category, and region. This suggests that increasing transaction volume or units sold does not automatically translate into improved profitability, particularly when transactions involve discounts or lower-margin products.

6. Product mix is a major differentiator in profitability

Profitability varies substantially across sub-categories. Binders and Copiers show statistically significant positive associations with profit, while Bookcases, Chairs, Machines, Storage, Supplies, and Tables show statistically significant negative associations relative to the regression's reference sub-category. Machines and Tables are among the strongest negative effects, indicating that product mix is an important consideration when evaluating profitability.

7. High-margin products provide important profit opportunities

Binders and Copiers demonstrate strong profitability characteristics and represent attractive areas for protecting margins and supporting profitable growth. The business should identify the characteristics of these successful product lines and consider whether similar strategies can be applied to other products without compromising demand.

8. Discounting can amplify weaknesses in low-margin products

The combination of descriptive analysis and regression results indicates that discounting is particularly problematic when applied to already weak product lines. Products with structurally lower profitability can experience further margin deterioration when discounts are increased. Discount decisions should therefore consider the underlying margin structure and profitability of individual products and sub-categories.

9. Profitability is influenced by multiple business factors

The OLS model explains approximately 33.7% of the variation in profit (R² = 0.337) and is statistically significant overall. The results indicate that profitability is influenced by a combination of sales, discounting, quantity, product sub-category, and regional factors rather than sales alone. However, the model does not explain all variation in profitability, so the findings should be interpreted as statistical associations rather than causal relationships.

10. Overall business implication
    
The central business challenge is not simply generating sales, but converting sales into sustainable profit. The analysis indicates that revenue growth can be undermined by excessive discounting, low-margin product lines, and inefficient sales volume. The strongest opportunity is therefore to improve the quality of revenue by protecting high-margin products, controlling excessive discounts, monitoring structurally unprofitable sub-categories, and evaluating growth using both sales and profitability measures.

## 18. Recommendations

1. Optimize Discounting Strategy

Limit excessive discounts, particularly on low-margin and loss-making product lines such as Tables, Machines, and other weaker sub-categories. Replace broad discounting with targeted promotions based on product profitability, customer behavior, and demand. The strong negative relationship between discount and profit in the regression reinforces the need for tighter discount controls.

2. Protect and Expand High-Margin Products

Prioritize profitable product opportunities, particularly high-performing sub-categories such as Binders and Copiers, while continuing to invest in strong categories such as Technology and Office Supplies. Management should identify the characteristics driving their profitability and use these insights to support profitable growth.

3. Reassess Low-Margin Product Lines

Conduct a detailed review of structurally weak sub-categories, particularly Tables, Machines, Bookcases, Chairs, Storage, and Supplies. Management should evaluate pricing, discount levels, product costs, and demand before deciding whether to redesign, reprice, reduce, or discontinue selected products.

4. Improve Shipping and Fulfillment Efficiency

Review shipping costs by product size, shipping mode, region, and order characteristics. Negotiate better carrier rates, explore order consolidation where practical, and evaluate whether expedited shipping options generate sufficient customer value to justify their additional cost. The objective should be to reduce fulfillment costs without compromising customer experience.

5. Develop Region-Specific Strategies

Use regional performance differences to develop targeted strategies rather than applying the same approach across all markets. Strong-performing regions such as the West and East should be protected and optimized, while weaker-performing regions such as the South should be investigated to identify whether pricing, product mix, discounting, demand, or distribution factors are limiting profitability.

6. Tailor Customer Segment Strategies

Continue prioritizing the Consumer segment because of its scale and contribution to overall sales and profit, while developing targeted strategies for Corporate customers. Although Home Office generates lower absolute sales and profit, its higher profit margin indicates that the segment should not simply be treated as underperforming. Instead, the business should investigate opportunities to increase its sales volume while protecting its relatively strong margin.

7. Implement Data-Driven Pricing and Promotion

Use historical sales, discount, product, customer, and profitability data to establish more disciplined pricing rules. Predictive and analytical models can be used to identify when discounts are likely to increase demand sufficiently to justify the reduction in margin, helping management move from broad discounting toward profit-aware promotional decisions.

8. Monitor Profitability Alongside Revenue Growth

Management should avoid evaluating business performance using revenue alone. Future performance monitoring should combine sales, profit, profit margin, discount rate, order volume, and loss-making orders to ensure that revenue growth is translating into sustainable financial performance.

There should be shift from revenue-focused growth to profit-focused growth. The analysis demonstrates that increasing sales alone does not guarantee improved profitability. Management should prioritize profitable revenue by controlling excessive discounts, protecting high-margin products, reassessing structurally unprofitable sub-categories, and using customer, regional, and product-level data to guide pricing and promotional decisions.
